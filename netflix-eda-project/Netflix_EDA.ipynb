```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df, x='year_added')

plt.title('Netflix Content Added by Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12, 5))

sns.histplot(data=df, x='release_year', bins=30)

plt.title('Distribution of Netflix Content by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df, x='rating', order=df['rating'].value_counts().index)

plt.title('Netflix Content by Rating')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(data=df, x='rating', hue='type')

plt.title('Ratings by Content Type')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)

plt.show()
```

```python
df[['type', 'duration']].head(10)
```

```python
movies = df[df['type'] == 'Movie'].copy()

movies['duration_min'] = movies['duration'].str.replace(' min', '', regex=False)

movies['duration_min'] = pd.to_numeric(movies['duration_min'], errors='coerce')
```

```python
movies[['title', 'duration', 'duration_min']].head(10)
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=movies, x='duration_min', bins=30)

plt.title('Distribution of Movie Durations')
plt.xlabel('Duration (minutes)')
plt.ylabel('Number of Movies')

plt.show()
```

```python
movies['duration_min'].describe()
```

```python
tv_shows = df[df['type'] == 'TV Show'].copy()
```

```python
tv_shows['seasons'] = tv_shows['duration'].str.replace(' Seasons', '', regex=False)
tv_shows['seasons'] = tv_shows['seasons'].str.replace(' Season', '', regex=False)

tv_shows['seasons'] = pd.to_numeric(tv_shows['seasons'], errors='coerce')
```

```python
tv_shows[['title', 'duration', 'seasons']].head(10)
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=tv_shows, x='seasons')

plt.title('Number of Seasons in Netflix TV Shows')
plt.xlabel('Number of Seasons')
plt.ylabel('Number of TV Shows')

plt.show()
```
